# Exploratory Data Analysis – Customer & Campaign Insights

## Objective
This analysis aims to understand:
- Geographic customer distribution
- Customer demographics
- Conversion patterns for new contracts
- Key segments driving campaign success


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("../output/cleaned_data.csv")
df.head()
```

## 1. Customer Distribution by Country

## 1. Customer Distribution by Country

```python
df["country"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    title="Customer Count by Country"
)
plt.ylabel("Number of Customers")
plt.show()
```

## 2. Top UK Towns by Customer Count


```python
df[df["country"] == "UK"]["town"].value_counts().head(10).plot(
    kind="bar",
    figsize=(9,5),
    title="Top 10 UK Towns by Customer Count"
)
plt.ylabel("Number of Customers")
plt.show()
```

## 4. Customer Age Distribution


```python
plt.figure(figsize=(8,5))
sns.histplot(df["age"], bins=20, kde=True)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Customer Age Distribution")
plt.show()
```

## 4. Job Type Distribution
This shows the employment categories represented in the dataset.



```python
df["job"].value_counts().head(10).plot(
    kind="bar",
    figsize=(9,5),
    title="Top Job Categories"
)
plt.ylabel("Number of Customers")
plt.show()
```

## 5. Education Level Distribution
This provides insight into customers' educational background.


```python
df["education"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    title="Education Level Distribution"
)
plt.ylabel("Number of Customers")
plt.show()
```

## 6. Marital Status Profile
This describes household and life-stage characteristics.


```python
df["married"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    title="Marital Status Distribution"
)
plt.ylabel("Number of Customers")
plt.show()
```

## 7. Communication Channel Used
This shows how customers were last contacted.


```python
df["last_contact"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    title="Last Contact Method"
)
plt.ylabel("Number of Customers")
plt.show()
```

## 8. Housing Status
This shows whether customers have housing-related commitments.


```python
df["housing"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    title="Housing Status Distribution"
)
plt.ylabel("Number of Customers")
plt.show()
```

## Key Dataset Insights
- The dataset is dominated by UK-based customers
- Customer presence is concentrated in major towns and cities
- Most customers fall within working-age groups
- Employment and education profiles are diverse
- The dataset represents a broad customer demographic base
- These insights support segmentation, targeting, and dashboard design
